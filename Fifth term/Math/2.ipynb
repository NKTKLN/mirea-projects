```python
def print_complex_forms(z, name="z"):
    print(f"{name} в алгебраической форме:     {z.real:.4f} + {z.imag:.4f}i")

    r = abs(z)
    theta = cmath.phase(z)
    print(f"{name} в тригонометрической форме: {r:.4f}(cos({theta:.4f}) + i·sin({theta:.4f}))")

    print(f"{name} в показательной форме:      {r:.4f} * e^(i*{theta:.4f})\n")


z_a = (5 + 1j) * (7 - 6j) / (3 + 1j)
print_complex_forms(z_a, "a)")

z_b = (1 + 1j) ** 5 / (1 - 1j) ** 3
print_complex_forms(z_b, "b)")

z_c = ((-1 + 1j * cmath.sqrt(3)) ** 15 / (1 - 1j) ** 20) + (
    (-1 - 1j * cmath.sqrt(3)) ** 15 / (1 + 1j) ** 20
)
print_complex_forms(z_c, "c)")
```

```python
def plot_roots(roots, title):
    plt.figure(figsize=(6, 6))

    for z in roots:
        plt.plot(z.real, z.imag, "ro")
        plt.text(z.real + 0.05, z.imag + 0.05, f"{z.real:.2f}+{z.imag:.2f}i", fontsize=8)

    plt.title(title)
    plt.axhline(0, color="black", linewidth=0.5)
    plt.axvline(0, color="black", linewidth=0.5)
    plt.show()
```

```python
R = 2 ** (1 / 16)
solutions = []

for k in range(8):
    theta = (cmath.pi / 32) + (cmath.pi / 4) * k
    z = R * cmath.exp(1j * theta)
    solutions.append(z)
    print(f"z{k} = {z.real:.4f} + {z.imag:.4f}i")

plot_roots(solutions, "Множество решений уравнения z^8 = 1 + i")
```

## 2.3

```python
def nth_roots(value, n):
    r, phi = cmath.polar(value)
    roots = []
    for k in range(n):
        root = r ** (1 / n) * cmath.exp(1j * (phi + 2 * cmath.pi * k) / n)
        roots.append(root)
    return roots


roots_a = nth_roots(1, 8)
print("a) 8√1:")
for i, z in enumerate(roots_a):
    print(f"z{i} = {z.real:.4f} + {z.imag:.4f}i")
plot_roots(roots_a, "a) Восьмые корни")

roots_b = nth_roots(-18 / (1 + 1j * 3**0.5), 4)
print("\nb) 4√(-18 / (1 + i√3)):")
for i, z in enumerate(roots_b):
    print(f"z{i} = {z.real:.4f} + {z.imag:.4f}i")
plot_roots(roots_b, "b) Четвёртые корни")

expr_c = (1 - 5j) / (1 + 1j - 5 * (1 + 2j) / (2 - 1j + 11))
roots_c = nth_roots(expr_c, 3)
print("\nc) 3√((1 - 5i)/(1 + i - 5*(1 + 2i)/(2 - i + 11))):")
for i, z in enumerate(roots_c):
    print(f"z{i} = {z.real:.4f} + {z.imag:.4f}i")
plot_roots(roots_c, "c) Кубические корни")

roots_d = nth_roots(1 - 1j, 9)
print("\nd) 9√(1 - i):")
for i, z in enumerate(roots_d):
    print(f"z{i} = {z.real:.4f} + {z.imag:.4f}i")
plot_roots(roots_d, "d) Девятые корни")
```

```python
def plot_mask(mask, title, x, y, color="lightblue"):
    plt.figure(figsize=(5, 5))
    plt.contourf(x, y, mask, levels=[0.5, 1], colors=[color])
    plt.axhline(0, color="black", linewidth=0.5)
    plt.axvline(0, color="black", linewidth=0.5)
    plt.title(title)
    plt.show()
```

```python
mask_a = np.abs(Z - 1j) + np.abs(Z + 1j) < 4
plot_mask(mask_a, "a) |z - i| + |z + i| < 4", X, Y, "lightblue")

mask_b = np.real(Z * (1 - 1j)) < np.sqrt(2)
plot_mask(mask_b, "b) Re(z(1 - i)) < √2", X, Y, "lightgreen")

mask_c1 = np.isclose(np.abs(Z_mini + 1), np.abs(Z_mini + 1j), atol=1e-1)
mask_c2 = np.isclose(np.abs(Z_mini + 1j), np.abs(Z_mini - 1j), atol=1e-1)
mask_c = mask_c1 & mask_c2
plot_mask(mask_c, "c) |z + 1| = |z + i| = |z - i|", X_mini, Y_mini, "salmon")

W = (1j - Z) / (Z + 1j)
argW = np.angle(W)
mask_d = (argW > 0) & (argW < np.pi / 2)
plot_mask(mask_d, "d) 0 < arg((i - z) / (z + i)) < π/2", X, Y, "orange")

W = (2 - 1j * Z) / (Z + 3j)
argW = np.angle(W)
mask_e = (argW <= -2 * np.pi / 3) & (argW >= -5 * np.pi / 6)
plot_mask(mask_e, "e) -5π/6 ≤ arg((2 - i z)/(z + 3i)) ≤ -2π/3", X, Y, "violet")

W = (3 - Z) / (Z + 1j)
argW = np.angle(W)
mask_f = (argW >= -2 * np.pi / 3) & (argW <= -np.pi / 3)
plot_mask(mask_f, "f) -2π/3 ≤ arg((3 - z) / (z + i)) ≤ -π/3", X, Y, "cyan")
```

```python
def julia_set(
    f,
    c=None,
    im_width=500,
    im_height=500,
    xmin=-2.1,
    xmax=2.1,
    ymin=-2.1,
    ymax=2.1,
    zabs_max=10,
    nit_max=1000,
):
    xwidth = xmax - xmin
    yheight = ymax - ymin
    julia = np.zeros((im_width, im_height))

    for ix in range(im_width):
        for iy in range(im_height):
            z = complex(ix / im_width * xwidth + xmin, iy / im_height * yheight + ymin)
            nit = 0
            while abs(z) <= zabs_max and nit < nit_max:
                if c is not None:
                    z = f(z, c)
                else:
                    z = f(z)
                nit += 1
            julia[ix, iy] = nit / nit_max
    return julia


def plot_julia(julia, win=5):
    plt.subplots(figsize=(8, 8))
    plt.grid(color="DarkTurquoise", alpha=0.75, linestyle=":", linewidth=0.5)
    plt.ylim(-win, win)
    plt.xlim(-win, win)
    plt.imshow(julia, cmap="flag", extent=[-win, win, -win, win])
    plt.axis("off")
    plt.show()
```

```python
# a) f_c(z) = z^2 + c
def f1(z, c):
    return z**2 + c


# б) f(z) = (i*z^5 - 1) / (z^5 - i)
def f2(z):
    return (1j * z**5 - 1) / (z**5 - 1j)


# в) f(z) = (z^2 + i) / (1 - z^2)
def f3(z):
    return (z**2 + 1j) / (1 - z**2)
```

```python
c1 = complex(-0.1, 0.65)
julia1 = julia_set(f1, c=c1)
plot_julia(julia1)

c2 = complex(0.28, 0.009)
julia2 = julia_set(f1, c=c2)
plot_julia(julia2)

julia3 = julia_set(f2)
plot_julia(julia3)

julia4 = julia_set(f3)
plot_julia(julia4)
```

```python
def stereographic_projection(z):
    x, y = z.real, z.imag
    denom = 1 + x**2 + y**2
    X = 2 * x / denom
    Y = 2 * y / denom
    Z = (-1 + x**2 + y**2) / denom
    return X, Y, Z


def plot_sphere(z_values, label=None):
    fig = plt.figure(figsize=(8, 8))
    ax = fig.add_subplot(111, projection="3d")

    u = np.linspace(0, 2 * np.pi, 100)
    v = np.linspace(0, np.pi, 100)
    X = np.outer(np.cos(u), np.sin(v))
    Y = np.outer(np.sin(u), np.sin(v))
    Z = np.outer(np.ones_like(u), np.cos(v))

    ax.plot_surface(X, Y, Z, color="lightgray", alpha=0.3, rstride=4, cstride=4, linewidth=0)
    ax.set_box_aspect([1, 1, 1])
    ax.plot(*stereographic_projection(z_values), color="red", lw=2, label=label)
    ax.set_title(label)
    plt.show()
```

```python
# Прямая: y = x
x = np.linspace(-5, 5, 400)
line = x + 1j * x

# Окружность: |z| = 2
theta = np.linspace(0, 2 * np.pi, 400)
circle = 2 * np.exp(1j * theta)

# Парабола: y = x^2
x_parab = np.linspace(-2, 2, 400)
parabola = x_parab + 1j * (x_parab**2)

# Гипербола: y = 1/x
x_hyper = np.linspace(-5, -0.1, 200)
x_hyper2 = np.linspace(0.1, 5, 200)
hyperbola = np.concatenate(
    [x_hyper + 1j * (1 / x_hyper), x_hyper2 + 1j * (1 / x_hyper2)]
)

# Логарифмическая спираль: r=ae^(bφ)
phi = np.linspace(0, 6 * np.pi, 1000)
r = 0.1 * np.exp(0.2 * phi)
log_spiral = r * np.exp(1j * phi)
```

```python
plot_sphere(line, "Прямая")
plot_sphere(circle, "Окружность")
plot_sphere(parabola, "Парабола")
plot_sphere(hyperbola, "Гипербола")
plot_sphere(log_spiral, "Логарифмическая спираль")
```
