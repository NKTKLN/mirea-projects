## 3 Задание

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.datasets import mnist
```

```python
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

train_images = np.pad(train_images, ((0, 0), (18, 18), (18, 18)), mode='constant')  # Добавляем нули по краям
train_images = train_images.reshape((60000, 64, 64, 1))

test_images = np.pad(test_images, ((0, 0), (18, 18), (18, 18)), mode='constant')
test_images = test_images.reshape((10000, 64, 64, 1))
```

```python
train_images = train_images.astype("float32") / 255
test_images = test_images.astype("float32") / 255
```

```python
inputs = keras.Input(shape=(64, 64, 1))
```

```python
x = layers.Conv2D(filters=4, kernel_size=3, activation='relu', padding='same')(inputs)  # Используем 'same' padding
x = layers.MaxPooling2D(pool_size=2)(x)

x = layers.Conv2D(filters=8, kernel_size=3, activation='relu', padding='same')(x)
x = layers.MaxPooling2D(pool_size=2)(x)

x = layers.Conv2D(filters=16, kernel_size=3, activation='relu', padding='same')(x)
x = layers.MaxPooling2D(pool_size=2)(x)

x = layers.Conv2D(filters=32, kernel_size=3, activation='relu', padding='same')(x)
x = layers.MaxPooling2D(pool_size=2)(x)

x = layers.Conv2D(filters=64, kernel_size=3, activation='relu', padding='same')(x)

x = layers.Flatten()(x)
outputs = layers.Dense(10, activation="softmax")(x)

model = keras.Model(inputs=inputs, outputs=outputs)
```

```python
model.compile(optimizer='rmsprop', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(train_images, train_labels, epochs=10, batch_size=32)
```

```python
loss, accuracy = model.evaluate(test_images, test_labels)
print(f'Loss: {loss}, Accuracy: {accuracy}')
```

## 4 Задание

```python
import numpy as np
from keras.datasets import cifar10
from keras.models import Sequential
from keras.layers import Dense, Flatten, Activation, Dropout
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from keras.utils import to_categorical
from keras.optimizers import SGD
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
(X_train, y_train), (X_test, y_test) = cifar10.load_data()
class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']
```

```python
np.random.seed(42)
plt.figure(figsize=(15,15))

for i in range(64):
  plt.subplot(8,8,i+1)
  plt.xticks([])
  plt.yticks([])
  plt.grid(False)
  plt.imshow(X_train[i])
  plt.title(class_names[y_train[i][0]], fontsize=12)
```

```python
X_train = X_train.astype('float32') / 255
X_test = X_test.astype('float32') / 255

Y_train = to_categorical(y_train, 10)
Y_test = to_categorical(y_test, 10)
```

```python
model = Sequential([
    # Первый каскад: Conv2D -> MaxPooling2D
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    Conv2D(32, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Dropout(0.25),

    # Второй каскад: Conv2D -> MaxPooling2D
    Conv2D(64, (3, 3), activation='relu'),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Dropout(0.25),

    # Преобразование в плоский формат
    Flatten(),

    # Полносвязные слои
    Dense(512, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')  # Выходной слой с 10 классами
])
```

```python
sgd = SGD(learning_rate = 0.01, momentum=0.0, nesterov=True)
model.compile(loss='categorical_crossentropy',
              optimizer=sgd,
              metrics=['accuracy'])

history = model.fit(X_train, Y_train, batch_size=32, epochs=25, validation_split=0.1, shuffle=True, verbose=1)
```

```python
loss, accuracy = model.evaluate(X_test, Y_test)
print(f'Loss: {loss}, Accuracy: {accuracy}')
```

#### Изменим гиперпараметры

```python
model = Sequential([
    # Первый каскад
    Conv2D(64, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Dropout(0.25),

    # Второй каскад
    Conv2D(128, (3, 3), activation='relu'),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Dropout(0.25),

    # Третий каскад
    Conv2D(256, (3, 3), activation='relu'),
    Conv2D(256, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Dropout(0.25),

    # Полносвязные слои
    Flatten(),
    Dense(1024, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])
```

```python
sgd = SGD(learning_rate=0.01, momentum=0.0, nesterov=True)
model.compile(loss='categorical_crossentropy', optimizer=sgd, metrics=['accuracy'])

history = model.fit(X_train, Y_train, batch_size=16, epochs=25, validation_split=0.1, shuffle=True, verbose=1)
```

```python
loss, accuracy = model.evaluate(X_test, Y_test)
print(f'Loss: {loss}, Accuracy: {accuracy}')
```
