4.3.2 Задание

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return (np.sqrt(1 + np.e ** np.sqrt(x) + np.cos(x ** 2)) / abs(1 - np.sin(x) ** 3)) + np.log(2 * x)
```

```python
x = np.arange(1, 11, 1)
y = function(x)

plt.grid()
plt.plot(x, y)
plt.scatter(x[:len(x)//2], y[:len(x)//2])
plt.show()
```

4.3.5 Задание

```python
import math as m

print("Чтобы выйти из программы введите exit")

while True:
    user_input = input("Введите (+, -, *, /, **, e, sin, cos): ")
    if user_input == "exit":
        break   
    elif user_input in ['+', '-', '*', '/', '**', 'e', 'sin', 'cos']:
        x = float(input("x = "))
        y = float(input("y = "))
        
        if user_input == 'e':
            print("e ** (x + y) = ", m.e**(x + y))
        elif user_input == 'sin':
            print("sin(x + y) =", m.sin(x + y))
        elif user_input == 'cos':
            print("cos(x + y) =", m.cos(x + y))
        else:
            formula = f"{x} {user_input} {y}"
            print(formula, "=", eval(formula))
    else:
        print("Неверный знак операции!")

print("Программа завершена")
```
