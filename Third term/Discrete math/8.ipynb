```python
import string
from collections import Counter
```

```python
RUSSIAN_FREQUENCIES = {
    'о': 0.1095,
    'е': 0.0849,
    'а': 0.0792,
    'и': 0.0731,
    'н': 0.0670,
    'т': 0.0628,
    'с': 0.0547,
    'л': 0.0451,
    'к': 0.0370,
    'м': 0.0324,
    'д': 0.0299,
    'п': 0.0280,
    'у': 0.0270,
    'г': 0.0162,
    'я': 0.0158,
    'ь': 0.0157,
    'з': 0.0147,
    'в': 0.0139,
    'ы': 0.0131,
    'б': 0.0097,
    'ч': 0.0096,
    'ж': 0.0075,
    'х': 0.0052,
    'й': 0.0051,
    'ш': 0.0049,
    'ц': 0.0044,
    'щ': 0.0026,
    'ъ': 0.0006,
    'э': 0.0005,
    'ф': 0.0004
}

ALPHABET = "абвгдеёжзийклмнопрстуфхцчшщъыьэюя"
```

```python
def caesar_cipher(alphabet: str, key: int, message: str) -> str:
    """Шифрует или дешифрует текст с помощью шифра Цезаря и заданного ключа."""
    new_message = ""
    for char in message:
        is_upper = char.isupper()
        char = char.lower()
        if char not in alphabet:
            new_message += char
            continue

        index_of_new_char = (alphabet.find(char) + key) % len(alphabet)
        new_char = alphabet[index_of_new_char]
        new_message += new_char.upper() if is_upper else new_char

    return new_message


def calculate_chi_squared(text: str, frequencies: dict[str, float]) -> float:
    """Рассчитывает значение хи-квадрат для частотной таблицы текста."""
    text_counts = Counter(filter(str.isalpha, text.lower()))
    chi_squared = 0
    total_chars = sum(text_counts.values())

    for letter, freq in frequencies.items():
        observed = text_counts.get(letter, 0)
        expected = total_chars * freq
        chi_squared += ((observed - expected) ** 2) / expected
    
    return chi_squared


def caesar_break(text: str, alphabet: str, frequencies: dict[str, float]) -> int:
    """Находит сдвиг, который минимизирует хи-квадрат, чтобы взломать шифр."""
    best_shift = 0
    min_chi_squared = float('inf')

    for shift in range(len(alphabet)):
        decrypted_text = caesar_cipher(alphabet, -shift, text)
        chi_squared = calculate_chi_squared(decrypted_text, frequencies)
        if chi_squared < min_chi_squared:
            min_chi_squared = chi_squared
            best_shift = shift

    return best_shift
```
