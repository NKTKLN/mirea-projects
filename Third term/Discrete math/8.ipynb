```python
import string
```

```python
def caesar_cipher(alphabet: str, key: int, message: str) -> str:
    new_message = ""
    for char in message:
        is_upper = char.isupper()
        char = char.lower()

        if char not in alphabet:
            new_message += char
            continue

        index_of_new_char = (alphabet.find(char) + key) % len(alphabet)
        new_char = alphabet[index_of_new_char]
        new_message += new_char.upper() if is_upper else new_char

    return new_message
```

### 2. Используя данные о вероятностном распределении символов русского языка, восстановить исходный текст (или его часть).

```python
message = message.lower()
message = message.translate(str.maketrans('', '', string.punctuation + '\n'))
```

```python
# Подсчет частоты букв в зашифрованном тексте
frequency_count = {}
for char in message:
    if char in alphabet:
        frequency_count[char] = frequency_count.get(char, 0) + 1

# Нормализуем частоты
total_letters = sum(frequency_count.values())
for char in frequency_count:
    frequency_count[char] /= total_letters

sorted_letters = sorted(frequency_count, key=frequency_count.get, reverse=True)
for char in sorted_letters:
    print(f"{char} | {frequency_count.get(char):.4f}")
```

```python
most_common_char = sorted_letters[0]

# Предполагаем, что наиболее частая буква в зашифрованном тексте соответствует 'о'
estimated_shift = (ord(most_common_char) - ord('о')) % 32

print("Ключ:", estimated_shift)
```

```python
decrypted_message = caesar_cipher(alphabet, -estimated_shift, message)
print("Расшифрованное сообщение:", decrypted_message)
```
