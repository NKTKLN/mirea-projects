```python
import random
```

### 1. Сгенерировать два больших простых числа

```python
def sieve_of_eratosthenes(limit):
    primes = [True] * (limit + 1)
    primes[0] = primes[1] = False  # 0 и 1 не простые числа
    p = 2
    while p * p <= limit:
        if primes[p]:
            for i in range(p * p, limit + 1, p):
                primes[i] = False
        p += 1
    return [p for p in range(2, limit + 1) if primes[p]]
```

```python
primes = sieve_of_eratosthenes(10000000)

p, q = random.sample(primes, 2)

print(f"Первое простое число p: {p}")
print(f"Второе простое число q: {q}")
```

### 2. Выполнить с числами операции сложения, вычитания, умножения, деления, вычисления остатка от деления одного числа на другое, возведения одного из чисел в степень n по модулю m.

```python
def modular_exponentiation(a, k, m):
    A = a
    B = 1
    while k > 0:
        if k % 2 == 1:
            B = (B * A) % m
        A = (A * A) % m
        k //= 2
    return B
```

```python
def karatsuba(x, y):
    if x < 10 or y < 10:
        return x * y

    m = min(len(str(x)), len(str(y))) // 2

    high1, low1 = divmod(x, 10**m)
    high2, low2 = divmod(y, 10**m)

    z0 = karatsuba(low1, low2)
    z1 = karatsuba(low1 + high1, low2 + high2)
    z2 = karatsuba(high1, high2)

    return z2 * 10**(2*m) + (z1 - z2 - z0) * 10**m + z0
```

```python
a = p + q
b = p - q
c = karatsuba(p, q)
d = p // q
mod_result = p % q

# 3. Возведение одного из чисел в степень n по модулю m
n = 5
m = primes[-3]
mod_exp = modular_exponentiation(p, n, m)

# Вывод результатов
print(f"Сложение: {a}")
print(f"Вычитание: {b}")
print(f"Умножение: {c}")
print(f"Целочисленное деление: {d}")
print(f"Остаток от деления: {mod_result}")
print(f"p в степени {n} по модулю {m}: {mod_exp}")
```
