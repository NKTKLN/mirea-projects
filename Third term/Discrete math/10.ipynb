```python
import numpy as np
```

```python
def generate_primes(k: int, m: int) -> list[int]:
    if m < 3:
        return None

    # Инициализация
    n = m + 2 * k - 2
    primes = [1 for _ in range(k)]  # Изначально считаем все числа в массиве простыми
    d = 3

    # Основной цикл для проверки делителей
    while d ** 2 <= n:
        r = m % d
        j = 1
        if r > 0 and r % 2 != 0:
            j += (d - r) // 2
        if m <= d:
            j += d

        # Вычеркиваем составные числа
        while j < k:
            primes[j] = 0
            j += d
        
        d += 4 if d % 6 == 1 else 2

    # Формируем список простых чисел
    result_primes = [m + 2 * i - 2 for i in range(k) if primes[i] == 1]

    return result_primes


big_prime_number_1 = generate_primes(100000, 104729)
big_prime_number_2 = generate_primes(1000, 1109)

print(big_prime_number_1)
print(big_prime_number_2)
```

### 1. Сгенерировать два больших простых числа

```python
def sieve_of_eratosthenes(limit):
    # Создаем булев массив
    is_prime = np.ones(limit + 1, dtype=bool)
    is_prime[0:2] = False

    for num in range(2, int(limit**0.5) + 1):
        if is_prime[num]:
            # Обнуляем все кратные num, начиная с его квадрата
            is_prime[num*num:limit+1:num] = False

    # Собираем все простые числа
    primes = np.nonzero(is_prime)[0]
    return primes
```

```python
primes = sieve_of_eratosthenes(100000000)

p = primes[-1]
q = primes[-2]

print(f"Первое простое число p: {p}")
print(f"Второе простое число q: {q}")
```

### 2. Выполнить с числами операции сложения, вычитания, умножения, деления, вычисления остатка от деления одного числа на другое, возведения одного из чисел в степень n по модулю m.

```python
def modular_exponentiation(a, k, m):
    A = a
    B = 1
    while k > 0:
        if k % 2 == 1:
            B = (B * A) % m
        A = (A * A) % m
        k //= 2
    return B
```

```python
a = p + q
b = p - q
c = p * q
d = p // q
mod_result = p % q

# 3. Возведение одного из чисел в степень n по модулю m
n = 5
m = primes[-3]
mod_exp = modular_exponentiation(p, n, m)

# Вывод результатов
print(f"Сложение: {a}")
print(f"Вычитание: {b}")
print(f"Умножение: {c}")
print(f"Целочисленное деление: {d}")
print(f"Остаток от деления: {mod_result}")
print(f"p в степени {n} по модулю {m}: {mod_exp}")
```
