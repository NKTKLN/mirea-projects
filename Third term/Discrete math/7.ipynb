```python
import string
from string import ascii_lowercase
```

### 1. Провести предварительную обработку произвольного текста. Привести все слова к нижнему регистру, удалить знаки препинания.

```python
text = "Lorem ipsum dolor sit amet, consectetur adipiscing elit."
```

```python
text = text.lower()  # Приведение к нижнему регистру
text = text.translate(str.maketrans('', '', string.punctuation + '\n'))  # Удаление знаков препинания
text = text.replace(" ", "")  # Удаление пробелов

print(text)
```

### 2. Зашифровать текст методом простой замены, запросив ключ шифрования у пользователя.

```python
def substitution_cipher(alphabet: str, replacement_alphabet: str, message: str) -> str:
    new_message = ""
    for char in message:
        is_upper = char.isupper()
        char = char.lower()

        if char not in alphabet:
            new_message += char
            continue

        new_char = replacement_alphabet[alphabet.find(char)]
        new_message += new_char.upper() if is_upper else new_char

    return new_message
```

### 2. Зашифровать текст методом простой замены, запросив ключ шифрования у пользователя.

```python
alphabet = ascii_lowercase
replacement_alphabet = "goydsipeluavcrjwxznhbqftmk"

encrypted_message = substitution_cipher(alphabet, replacement_alphabet, text)
print("Зашифрованное сообщение:", encrypted_message)
```

### 3. Дешифровать текст, используя тот же самый ключ.

```python
decrypted_message = substitution_cipher(replacement_alphabet, alphabet, encrypted_message)
print("Расшифрованное сообщение:", decrypted_message)
```

### 4. Зашифровать текст методом сложной замены, запросив ключ шифрования у пользователя.

```python
def vigenere(plaintext, key, is_encrypt=True):
    ciphertext = []
    key_length = len(key)
    key_as_int = [ord(i) - ord('a') for i in key]
    plaintext_int = [ord(i) - ord('a') for i in plaintext]

    for i in range(len(plaintext_int)):
        value = (plaintext_int[i] - key_as_int[i % key_length] * ((-1)**int(is_encrypt))) % 26
        ciphertext.append(chr(value + ord('a')))

    return ''.join(ciphertext)
```

```python
key = "hello"

encrypted_message = vigenere(text, key)
print("Зашифрованное сообщение:", encrypted_message)
```

### 5. Дешифровать текст, используя тот же самый ключ.

```python
decrypted_message = vigenere(encrypted_message, key, is_encrypt=False)
print("Расшифрованное сообщение:", decrypted_message)
```
