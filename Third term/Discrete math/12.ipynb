### 1. Выполнить шифрование и расшифровку открытого текста с помощью выбранной библиотеки

```python
message = "Lorem ipsum dolor sit amet, consectetur adipiscing elit."
```

```python
message = message.encode("utf8")
```

```python
# Генерируем приватный и публичный ключи
pubkey, privkey = rsa.newkeys(1024)

print(f"Приватный ключ: {pubkey}")
print(f"Публичный ключ: {privkey}")
```

```python
encrypted_message = rsa.encrypt(message, pubkey)
print("Зашифрованное сообщение:", encrypted_message)
```

```python
decrypted_message = rsa.decrypt(encrypted_message, privkey)
print("Расшифрованное сообщение:", decrypted_message)
```
