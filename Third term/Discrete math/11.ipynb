```python
from random import choice
```

```python
def gcd(a: int, b: int) -> int:
    if b == 0:
        return a

    return gcd(b, a % b)
```

```python
def gcd_extended(a: int, b: int) -> tuple[int, int, int]:
    if b == 0:
        return a, 1, 0

    gcd, x, y = gcd_extended(b, a % b)
    return gcd, y, x - (a // b) * y
```

```python
def mod_inverse(n: int, x: int) -> int:
    inv_x = gcd_extended(n, x)[2]
    if inv_x <= 0:
        inv_x += n
    return inv_x
```

```python
def mod_extended(x: int, y: int, n: int) -> int:
    if y == 0:
        return 1

    z = mod_extended(x, y // 2, n)

    if y % 2 == 0:
        return (z * z) % n

    return (x * z * z) % n
```

```python
def get_primes(n: int):
    sieve = [True] * n
    primes = []
    for p in range(2, n):
        if sieve[p]:
            primes.append(p)
            for i in range(p * p, n, p):
                sieve[i] = False
    return primes
```

### 1. Сгенерировать два простых числа. Получить открытый и закрытый ключи алгоритма RSA.

```python
primes = get_primes(100000)
p = choice(primes)
q = choice(primes)
while p == q:
    q = choice(primes)

n = p * q
phi = (p - 1) * (q - 1)

e = 3
for ferm_number in [5, 17, 257, 65537]:
    if ferm_number < phi:
        e = ferm_number

d = mod_inverse(phi, e)

public, private = (e, n), (d, n)

print(f"Приватный ключ: {public}")
print(f"Публичный ключ: {private}")
```

### 2. Подготовить открытый текст для шифрования и разбить его на блоки.

```python
message = "Lorem ipsum dolor sit amet, consectetur adipiscing elit."
```

```python
splited_message = [ord(char) for char in message]

print(*splited_message)
```

### 3. Провести шифрование текста

```python
e, n = public
encrypted_message = [mod_extended(char, e, n) for char in splited_message]
print("Зашифрованное сообщение:", "".join(map(str, encrypted_message)))
print(encrypted_message)
```

### 4. Расшифровать текст

```python
d, n = private
decrypted_message = [chr(mod_extended(char, d, n)) for char in encrypted_message]
print("Расшифрованное сообщение:", "".join(decrypted_message))
```
